```python
# Import log from file
nmea_log = NMEALog.from_file("trackWA1_2025-2026.nmea")
```

```python
start_point_lat = nmea_log.gga_sentences[0].latitude
start_point_lon = nmea_log.gga_sentences[0].longitude
start_point_alt = nmea_log.gga_sentences[0].get_alt_above_ellipsoid()
start_coords = WGS84Coords(start_point_lat, start_point_lon, start_point_alt)

end_point = nmea_log.gga_sentences[-1]
end_point_lat = end_point.latitude
end_point_lon = end_point.longitude
end_point_alt = end_point.get_alt_above_ellipsoid()
end_coords = WGS84Coords(end_point_lat, end_point_lon, end_point_alt)



# a)
print(start_coords.to_dms_string())

# b)
print(start_coords.to_ecef_string())

# c)
print(end_coords.to_dms_string())
az, el = start_coords.az_el_to(end_coords)
e, n, u = start_coords.enu_to(end_coords)
az2, el2 = enu_to_az_el(e, n, u)
print(f"Azimuth: {az}, elevation: {el}")
print(f"Azimuth: {az2}, elevation: {el2}")

# d)
min_alt_sent = nmea_log.get_sentence_by_attr_min("altitude")
print(min_alt_sent.string)
print(min_alt_sent.time)

# e)
total_distance = nmea_log.get_distance_travelled()
print(total_distance)
```
