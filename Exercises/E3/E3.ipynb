NB: Professoren oppga parametre med minus foran, men siden jeg går fra WGS84 til EU1950 så skal de være positive. I oppgaven står det 1950 til WGS84. Husk og sjekk dette.

```python
# Molodensky transformation
da = 251
df = 0.14192702E-4
dx = 84
dy = 107
dz = 120   

p1_eu1950 = p1.molodensky_transform(da, df, dx, dy, dz)
p2_eu1950 = p2.molodensky_transform(da, df, dx, dy, dz)
print(p1_eu1950.to_decimal_string(8))
print(p2_eu1950.to_decimal_string(8))

p1_translated = p1.translate_ecef(-dx, -dy, -dz)
p2_translated = p2.translate_ecef(-dx, -dy, -dz)
print(p1_translated)
print(p2_translated)
```
